# Poker equity in a notebook

Named players, reusable ranges, and an HTML result table. All calculations use the Rust extension.

```python
from poker_engine import Calculator, Range, evaluate_hand
```

```python
holdem = Calculator({'Hero': 'TT+,AQs+,AKo', 'Villain': 'random'}, board='2c3d4h')
result = holdem.simulate(samples=100_000, seed=42, threads=4)
result
```

```python
# The context shares parsed ranges; optional pandas display:
turn = holdem.with_board('2c3d4h8s')
turn.simulate(samples=100_000, seed=42, threads=4).to_frame()
```

```python
omaha = Calculator({'Hero': 'AA$ds', 'Villain': 'AKQJ$ds'}, game='omaha')
omaha.simulate(samples=100_000, seed=42, threads=4)
```

```python
split = Calculator({'Hero': 'As2cJhJd', 'Villain': 'Ah2dKsKh'}, game='omaha8', board='3d4h8sJcQd')
split.exact()  # Hero wins high, both tie low: 75% / 25%
```

```python
weighted = Range.from_hands({'AsAh': 2.0, 'KsKh': 1.0})
Calculator({'Hero': weighted, 'Villain': 'QQ'}, board='2c3d4h8s9c').exact()
```

```python
evaluate_hand('As2cKdKh', '3d4h5sQcQd', game='omaha8')
```

Install `poker-engine[notebook]` to use `to_frame()`. Plain results and `rows()` work without pandas. Seeded runs reproduce with the same thread count; changing threads changes the stream.
